```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/train.csv")
df
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
df
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df
```

```python
df["Sex"].value_counts()
```

```python
df["Survived"].value_counts()
```

```python
sns.pairplot(df[["Age","Sex","Fare","Parch"]].dropna())
plt.show()
```

The pairplot displays the distributions of individual numerical variables along the diagonal and scatterplots for pairs of variables elsewhere. It helps identify possible associations, clusters, skewed distributions, and unusual observations. The strength and direction of any relationship should be assessed from the actual scatterplots.

```python
numeric_df=df.select_dtypes(include="integer")
correlation=numeric_df.corr()

plt.figure(figsize=(10,10))
sns.heatmap(correlation,annot=True)
plt.show()
```

The heatmap visualizes the correlation coefficients between numerical variables. Positive values indicate positive linear associations, while negative values indicate negative linear associations. Values close to zero suggest weak linear associations. The displayed coefficients help identify which variable pairs have the strongest relationships. Correlation does not imply causation.

```python
df.groupby("Sex")["Survived"].mean()*100
```

**The survival rate differed between male and female passengers. The group with the higher survival rate had a greater proportion of survivors in this dataset.**

```python
sns.barplot(data=df, x="Sex", y="Survived")
plt.title("Survival Rate by Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.scatterplot(data=df, x="Fare", y="Parch")
plt.title("Relationship Between Fare and Parch")
plt.xlabel("Fare")
plt.ylabel("Number of Parents/Children")
plt.show()
```

The scatterplot explores the relationship between ticket fare and the number of parents or children accompanying each passenger. The distribution of points can reveal possible patterns, clusters, or unusually high fares. Because Parch takes discrete integer values, overlapping points are expected. A clear relationship should not be assumed unless the plotted data supports it.

**The age distribution shows which age ranges are most common and whether passenger ages are concentrated in particular intervals.**

```python
sns.histplot(data=df, x="Age", bins=20, kde=True)
plt.title("Distribution of Passenger Ages")
plt.show()
```

The histogram shows the distribution of passenger ages. It helps identify the most common age ranges and shows whether the distribution is symmetrical or skewed. The shape of the histogram also helps reveal whether very young or older passengers were relatively uncommon.

```python
sns.barplot(data=df, x="Pclass", y="Survived")
plt.title("Survival Rate by Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

The visualization compares survival rates across first-, second-, and third-class passengers. Differences in bar heights indicate differences in the proportion of passengers who survived across the classes. The class with the highest bar had the highest observed survival rate.

```python
sns.boxplot(data=df, x="Fare")
plt.title("Distribution of Ticket Fares")
plt.show()
```

The boxplot summarizes the distribution of ticket fares using the median, quartiles, and whiskers. Points beyond the whiskers indicate potential outliers. If the upper whisker is longer or there are many high-fare outliers, this suggests that the fare distribution is positively skewed.

## Summary of Findings

The Exploratory Data Analysis (EDA) was performed on the Titanic dataset using Pandas, Matplotlib, and Seaborn. Statistical summaries and visualizations were used to understand the distribution of variables, compare passenger groups, identify potential outliers, and explore relationships between numerical and categorical variables.

Histograms and boxplots helped examine distributions and potential outliers. Count plots and bar plots helped compare passenger categories and survival rates. Scatterplots, pairplots, and correlation heatmaps were used to investigate relationships between variables.

The analysis provided an overview of the dataset's characteristics and helped identify patterns that could be explored further. The conclusions were based on the observed graphs and calculated statistics, while recognizing that correlation does not establish causation.
